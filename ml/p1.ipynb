```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Supongamos que cargamos un dataset real (ej. de Kaggle: https://www.kaggle.com/datasets/pablobravo73/real-estate-bogota)
# df = pd.read_csv('inmuebles_bogota.csv')
# Para ejemplo, creamos uno pequeño representativo

data = {
    'tipo': ['apartamento', 'casa', 'apartamento', 'casa', 'oficina', 'apartamento'],
    'barrio': ['Chapinero', 'Usaquén', 'Kennedy', 'Suba', 'Chico', 'Teusaquillo'],
    'habitaciones': [3, 4, 2, 5, np.nan, 1],
    'area_m2': [85.5, 180.0, 60.2, 250.0, 120.0, 45.0],
    'precio_millones': [450, 1200, 220, 950, 800, 180],
    'estrato': [4, 6, 2, 5, 5, 3],
    'banos': [2, 3, 1, 4, 2, 1]
}
df = pd.DataFrame(data)

print("Forma:", df.shape)
print("\nInfo general:")
print(df.info())
print("\nPrimeras filas:")
print(df.head())
```

```python
# Perfilamiento básico
print("Registros y columnas:", df.shape)
print("\nTipos de datos:\n", df.dtypes)
print("\nValores únicos por columna:\n", df.nunique())

# Asociaciones iniciales (cuantitativas)
print("\nCorrelación entre variables numéricas:")
print(df.select_dtypes(include=['float64','int64']).corr())

# Subconjuntos interesantes: precio promedio por barrio
print("\nPrecio promedio por barrio:")
print(df.groupby('barrio')['precio_millones'].mean().sort_values(ascending=False))
```

```python
# Corregir tipos (ejemplo: estrato a ordinal/categórico)
df['estrato'] = df['estrato'].astype('category')  # o 'int' si no hay nulos
df['tipo'] = df['tipo'].astype('category')
df['barrio'] = df['barrio'].astype('category')

# Atención a numéricos que son cualitativos (ej. si estrato estuviera como float por error)
print(df.dtypes)

# Medidas según tipo
# Cuantitativos: tendencia central + dispersión
print("\nEstadísticas cuantitativas:")
print(df.describe())  # mean, std, min, max, etc. para area y precio

# Cualitativos: frecuencias
print("\nFrecuencias cualitativas:")
print(df['tipo'].value_counts(normalize=True) * 100)  # porcentajes
print(df['estrato'].value_counts().sort_index())
```

```python
# Completitud
print("Valores nulos por columna:\n", df.isnull().sum())

# Unicidad (duplicados totales)
print("Filas duplicadas:", df.duplicated().sum())

# Validez (ejemplos)
print("Áreas inválidas (negativas o extremas):")
print(df[(df['area_m2'] <= 0) | (df['area_m2'] > 1000)])  # rango razonable

print("Precios inválidos:")
print(df[df['precio_millones'] < 50])  # muy bajo para Bogotá

# Consistencia (ej. estrato solo 1-6)
```

```python
# Histograma + boxplot cuantitativos
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.histplot(df['precio_millones'], kde=True)
plt.title('Distribución de Precios (millones COP)')

plt.subplot(1,2,2)
sns.boxplot(y=df['precio_millones'])
plt.title('Boxplot Precios - detección outliers')
plt.show()

# Boxplot por categoría (asociación cuantitativo-cualitativo)
plt.figure(figsize=(10,6))
sns.boxplot(x='tipo', y='precio_millones', data=df)
plt.title('Precio por Tipo de Inmueble')
plt.xticks(rotation=45)
plt.show()

# Diagrama de dispersión (relación numéricas)
plt.figure(figsize=(8,6))
sns.scatterplot(x='area_m2', y='precio_millones', hue='estrato', size='habitaciones', data=df)
plt.title('Relación Área vs Precio (color=estrato, tamaño=habitaciones)')
plt.show()

# Mapa de calor de correlaciones
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm')
plt.title('Mapa de Calor - Correlaciones')
plt.show()

# Diagrama de torta (cualitativo)
plt.figure(figsize=(6,6))
df['tipo'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title('Distribución por Tipo de Propiedad')
plt.show()
```
